For convenience always start with new line as a  data marker. Data markers get added to the system prompt. 
Also called spotlighting. The below uses an XML and JSON data marker 

Data markers help an LLM separate instruction from data 

```python
!spikee --quiet generate \
    --seed-folder datasets/seeds-cybersec \
    --format full-prompt \
    --include-system-message \
    --spotlighting-data-markers $'\n<data>\nDOCUMENT\n<\data>\n',$'\n{"document":"DOCUMENT"}\n' \
    --instruction-filter data-exfil-markdown \
    --languages en
```
