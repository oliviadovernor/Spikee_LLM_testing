Test 1: Attack on standalone LLM

In our prompt injection methodology we have 3 phases:
- First we generate a payload, dataset /list of payload
- Perform attack
- Analsye results 

If we are looking at a target = standalone LLM attack, the attacker will control the whole payload/entire prompt that is submitted (including system message, delimiters, task, suffixes and user content).

The tasks can be any e.g. summarise and Q+A

Stage 1: creating the dataset. We can apply certain filters e.g. data exfiltration and english language to reduce the dataset. ! runs bash, you could also run in the terminal.

2 Datasets have been created 'cybersec-full-prompt-dataset-1791112471.jsonl' and 'cybersec-user-input-dataset-1791111907.jsonl'. One is a more filtered version of the other to test with.

2. Perform the attack. 
Free API tier is rate limiting so have a backoff implemented = 8000 tokens/min is Groq so need to be less than 133 tokens/second



```python
import subprocess
import time

while True:
    result = subprocess.run([
        "spikee", "--quiet", "test",
        "--dataset", "datasets/cybersec-full-prompt-dataset-gd-1791128511.jsonl",
        "--target", "llm_provider",
        "--target-options", "model=groq/openai/gpt-oss-20b",
        "--auto-resume"
    ])
    
    if result.returncode == 0:
        print("Done!")
        break
    
    print("Failed, retrying in 30s...")
    time.sleep(30)
```
